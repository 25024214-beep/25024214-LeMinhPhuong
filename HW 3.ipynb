### Bài 3: Nhập vào số nguyên dương a, in ra bảng cửu chương của a

### Bài 4: Viết chương trình in ra hình tam giác cân có độ cao h được nhập từ bàn phím h=int(input())

```python
h = int(input())
for i in range(1, h+1):
    print(" " * (h-i), end="")
    print("*" * (2*i-1))
```

### Bài 5: Nhập vào số nguyên dương n. Tính S = 1 + 2 + 3 + 4 + … + n

```python
n = int(input())
S = 0
for i in range(1, n+1):
    S += i
print(S)
```

### Bài 6: Nhập vào số nguyên dương a, in ra tất cả các ước dương của a

```python
a = int(input())
for i in range(1, a+1):
    if a % i == 0:
        print(i, end=" ")
```

### Bài 7: Nhập vào số nguyên dương a, đếm số ước dương của a

```python
a = int(input())
count = 0
for i in range(1, a+1):
    if a % i == 0:
        count += 1
print(count)
```

### Bài 8: Nhập vào số nguyên dương a và b, in toàn bộ ước dương chung của a và b

```python
a, b = map(int, input().split())

for i in range(1, min(a, b) + 1):
    if a % i == 0 and b % i == 0:
        print(i, end=" ")
```

### Bài 9: Nhập vào số nguyên dương a, kiểm tra xem a có phải là số chính phương hay không? Số chính phương là số có căn bậc hai là một số nguyên VD: 1, 4, 9, 16, 25…. 

```python
a = int(input())
if int(a**0.5)**2 == a:
    print("True")
else:
    print("False") 
```

### Bài 10: 
Nhập vào số nguyên dương a, kiểm tra xem a có phải là số hoàn hảo hay không. 
Số hoàn hảo là số có tổng các ước (ước nhỏ hơn nó) bằng chính nó, VD: số 6, 28… 

```python
a = int(input())
tong = 0
for i in range(1, a):
    if a % i == 0:
        tong += i
if tong == a:
    print("True")
else:
    print("False")

```
