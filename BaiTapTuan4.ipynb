### W4A11 
[EvenDivisors] 
Viết chương trình đếm số lượng ước số chẵn của số nguyên dương n với n < 10^6.

```python
n = int(input())
count = 0

for i in range(1, n + 1):
    if n % i == 0 and i % 2 == 0:
        count += 1

print(count)
```
